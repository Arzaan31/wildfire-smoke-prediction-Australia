Fix the "24:00" hour format and create proper datetime column.

WHY: Raw air quality data uses "24:00" as the last hour of the day,
but this is invalid for datetime parsing. We convert it to "00:00"
of the NEXT day. Correct datetime is essential for all temporal
operations: lags, rolling windows, merges with hourly hotspot data.


```python
df_aq["station"] = df_aq["station"].str.strip().str.title()

# Drop rows where Date or Time is missing (prevents parsing errors)
df_aq = df_aq.dropna(subset=["Date", "Time"])

# Fix '24:00' / '24:00:00' format
time_str = df_aq["Time"].astype(str).str.strip()
is_24_hr = time_str.str.startswith("24:")
cleaned_time = time_str.copy()
cleaned_time[is_24_hr] = cleaned_time[is_24_hr].str.replace(r"^24:", "00:", regex=True)

# Combine Date + Time into datetime
df_aq["datetime"] = pd.to_datetime(
    df_aq["Date"].astype(str) + " " + cleaned_time,
    format="mixed", dayfirst=True
)

# Add 1 day to rows that originally had hour "24:"
df_aq.loc[is_24_hr, "datetime"] += pd.Timedelta(days=1)

# Sort chronologically per station
df_aq = df_aq.sort_values(["station", "datetime"]).reset_index(drop=True)

# Drop original Date and Time columns (we now have datetime)
df_aq = df_aq.drop(columns=["Date", "Time"], errors="ignore")

print(f"Air Quality after datetime fix: {df_aq.shape}")
print(f"\nPer-station summary:")
for stn in sorted(df_aq["station"].unique()):
    s = df_aq[df_aq["station"] == stn]
    pm_valid = s["pm25"].notna().sum()
    temp_valid = s["temp"].notna().sum()
    print(f"  {stn:15s}: {len(s):6d} rows | PM2.5 valid: {pm_valid:6d} | "
          f"Temp valid: {temp_valid:6d}")
```

Scientifically correct imputation — different strategies per variable:

PM2.5 (target):
  - Interpolate gaps ≤3 hours only
  - Large gaps stay NaN → dropped later
  - WHY: Never fabricate target data for long periods

Temperature & Humidity:
  - Interpolate ≤12 hours — these change slowly
  
Wind Speed:
  - Interpolate ≤12 hours

Wind Direction:
  - Forward fill (ffill) ONLY — NOT interpolation
  - WHY: Direction is angular. Interpolating 350°→10° gives 180° (completely wrong!)
  - e.g., 270, NaN, NaN, 280 → 270, 270, 270, 280

Rainfall:
  - Create binary flag 'rainfall_missing' FIRST
  - Then fill NaN with 0
  - WHY: Lets models distinguish real zero rain from sensor-offline


```python
print("Missing values BEFORE cleaning:")
for stn in sorted(df_aq["station"].unique()):
    s = df_aq[df_aq["station"] == stn]
    print(f"  {stn:15s}: pm25={s['pm25'].isna().sum():5d}, "
          f"temp={s['temp'].isna().sum():5d}, "
          f"humidity={s['humidity'].isna().sum():5d}, "
          f"wind_dir={s['wind_direction'].isna().sum():5d}")

# Process each station SEPARATELY to avoid cross-station interpolation
cleaned_frames = []
for station_name in sorted(df_aq["station"].unique()):
    stn = df_aq[df_aq["station"] == station_name].copy()
    stn = stn.sort_values("datetime").reset_index(drop=True)

    # PM2.5: interpolate small gaps only (≤3 hours)
    stn["pm25"] = stn["pm25"].interpolate(method="linear", limit=3)

    # Temperature: interpolate ≤12 hours
    stn["temp"] = stn["temp"].interpolate(method="linear", limit=12)

    # Humidity: interpolate ≤12 hours
    stn["humidity"] = stn["humidity"].interpolate(method="linear", limit=12)

    # Wind Speed: interpolate ≤12 hours
    stn["wind_speed"] = stn["wind_speed"].interpolate(method="linear", limit=12)

    # Wind Direction: forward fill only (NOT interpolation)
    stn["wind_direction"] = stn["wind_direction"].ffill()

    # Rainfall: create missing flag THEN fill with 0
    stn["rainfall_missing"] = stn["rainfall"].isna().astype(int)
    stn["rainfall"] = stn["rainfall"].fillna(0)

    cleaned_frames.append(stn)

df_aq = pd.concat(cleaned_frames, ignore_index=True)

print("\nMissing values AFTER cleaning:")
for stn in sorted(df_aq["station"].unique()):
    s = df_aq[df_aq["station"] == stn]
    print(f"  {stn:15s}: pm25={s['pm25'].isna().sum():5d}, "
          f"temp={s['temp'].isna().sum():5d}, "
          f"humidity={s['humidity'].isna().sum():5d}, "
          f"wind_dir={s['wind_direction'].isna().sum():5d}")
```

Vectorized geographic calculations for fire-exposure features.

Three functions:
  1. haversine_vectorized — Fast distance from station to ALL fires (numpy)
  2. calculate_bearing_vectorized — Compass bearing FROM station TO each fire
  3. compute_fire_features_for_window — ALL spatial features for one station-hour:
     - nearest_fire_km
     - fires_50km, fires_100km, fires_200km, fires_500km (counts within radii)
     - frp_50km, frp_100km, frp_200km, frp_500km (FRP sums within radii)
     - fire_exposure = Σ(FRP / distance²) — inverse square law
     - upwind_exposure = Σ(FRP × cos(wind_angle_diff) / distance²)

WHY upwind_exposure is the most physically meaningful fire variable:
  It combines three physical factors:
    INTENSITY (FRP) — larger fires produce more smoke
    DISTANCE (1/d²) — smoke disperses with distance squared
    WIND DIRECTION — fires UPWIND affect you, fires downwind don't
  A 100MW fire upwind at 50km is far more dangerous than a 100MW fire
  downwind at 10km. Simple fire count ignores all of this physics.


```python
def haversine_vectorized(lat1, lon1, lat2_arr, lon2_arr):
    """
    Vectorized Haversine distance calculation.
    lat1, lon1: scalar (station coordinates)
    lat2_arr, lon2_arr: numpy arrays (fire coordinates)
    Returns: distances in km
    """
    R = 6371.0  # Earth's radius in km
    lat1_r = np.radians(lat1)
    lat2_r = np.radians(lat2_arr)
    dlat = np.radians(lat2_arr - lat1)
    dlon = np.radians(lon2_arr - lon1)
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1_r) * np.cos(lat2_r) * np.sin(dlon / 2) ** 2
    c = 2 * np.arcsin(np.sqrt(np.clip(a, 0, 1)))
    return R * c


def calculate_bearing_vectorized(lat1, lon1, lat2_arr, lon2_arr):
    """
    Vectorized bearing FROM station TO fires.
    Returns bearings in degrees (0–360).
    """
    lat1_r = np.radians(lat1)
    lat2_r = np.radians(lat2_arr)
    dlon = np.radians(lon2_arr - lon1)
    x = np.sin(dlon) * np.cos(lat2_r)
    y = np.cos(lat1_r) * np.sin(lat2_r) - np.sin(lat1_r) * np.cos(lat2_r) * np.cos(dlon)
    bearing = np.degrees(np.arctan2(x, y))
    return (bearing + 360) % 360


def compute_fire_features_for_window(fire_lats, fire_lons, fire_powers,
                                      st_lat, st_lon, wind_dir):
    """
    Compute all spatial fire features for a given set of fires
    relative to one monitoring station at one time point.
    """
    result = {
        "nearest_fire_km": 999.0,
        "fires_50km": 0, "fires_100km": 0, "fires_200km": 0, "fires_500km": 0,
        "frp_50km": 0.0, "frp_100km": 0.0, "frp_200km": 0.0, "frp_500km": 0.0,
        "fire_exposure": 0.0,
        "upwind_exposure": 0.0,
    }

    if len(fire_lats) == 0:
        return result

    # Vectorized distance and bearing
    distances = haversine_vectorized(st_lat, st_lon, fire_lats, fire_lons)
    bearings = calculate_bearing_vectorized(st_lat, st_lon, fire_lats, fire_lons)
    powers = fire_powers

    # Nearest fire
    result["nearest_fire_km"] = float(np.min(distances))

    # Counts within radii
    for radius, key in [(50, "fires_50km"), (100, "fires_100km"),
                         (200, "fires_200km"), (500, "fires_500km")]:
        result[key] = int(np.sum(distances <= radius))

    # FRP sums within radii
    for radius, key in [(50, "frp_50km"), (100, "frp_100km"),
                         (200, "frp_200km"), (500, "frp_500km")]:
        result[key] = float(np.sum(powers[distances <= radius]))

    # Fire Exposure Index = Σ(FRP / distance²)
    valid = distances > 0
    if np.any(valid):
        result["fire_exposure"] = float(np.sum(powers[valid] / (distances[valid] ** 2)))

    # Upwind Exposure Index = Σ(FRP × cos(angle_diff) / distance²)
    if not np.isnan(wind_dir) and np.any(valid):
        angle_diff = np.abs(bearings[valid] - wind_dir)
        angle_diff = np.minimum(angle_diff, 360 - angle_diff)
        alignment = np.maximum(0, np.cos(np.radians(angle_diff)))
        result["upwind_exposure"] = float(
            np.sum(powers[valid] * alignment / (distances[valid] ** 2))
        )

    return result


# Pre-index hotspots by date for fast window lookups
df_hotspots["date"] = df_hotspots["datetime"].dt.date
hotspot_date_index = df_hotspots.groupby("date").apply(
    lambda g: g[["latitude", "longitude", "power", "datetime"]].values
).to_dict()


def get_fires_in_window(t_start, t_end):
    """
    Efficiently retrieve all fires in [t_start, t_end] time window.
    Uses date-based pre-indexing for speed.
    """
    start_date = t_start.date()
    end_date = t_end.date()

    fires_list = []
    current_date = start_date
    while current_date <= end_date:
        if current_date in hotspot_date_index:
            fires_list.append(hotspot_date_index[current_date])
        current_date += pd.Timedelta(days=1)

    if not fires_list:
        return np.array([]), np.array([]), np.array([])

    all_fires = np.vstack(fires_list)
    fire_datetimes = pd.to_datetime(all_fires[:, 3])
    mask = (fire_datetimes >= t_start) & (fire_datetimes <= t_end)

    if not np.any(mask):
        return np.array([]), np.array([]), np.array([])

    filtered = all_fires[mask]
    return (
        filtered[:, 0].astype(float),  # latitudes
        filtered[:, 1].astype(float),  # longitudes
        filtered[:, 2].astype(float),  # powers (FRP)
    )


print(f"✅ Spatial functions ready. Hotspot date index: {len(hotspot_date_index)} dates.")
```

Main fire feature computation loop.

For EVERY station-hour observation, we look at fires within a 24-HOUR
LOOKBACK WINDOW and compute spatial fire features.

WHY 24-HOUR LOOKBACK:
  Smoke from fires doesn't arrive instantly. A fire burning 200km away
  12 hours ago can still affect current air quality through atmospheric
  transport. Using only same-hour fires would miss the majority of
  smoke impacts. 24 hours captures the typical transport timescale
  for distances up to ~500km in typical Australian wind conditions.

PROCESS per station:
  1. Get station coordinates from Stations.csv
  2. For each hourly observation:
     a. Define window [t-24h, t]
     b. Retrieve all fires in that window
     c. Compute distances, bearings, FRP-weighted features
     d. Store results
  3. Save per-station Parquet file

RUNTIME: ~2–5 minutes per station depending on data size.


```python

LOOKBACK_HOURS = 24

station_list = sorted(df_aq["station"].unique())
print(f"Processing {len(station_list)} stations: {station_list}\n")

for station_name in station_list:

    print(f"{'=' * 60}")
    print(f"PROCESSING: {station_name}")
    print(f"{'=' * 60}")

    # Get station coordinates
    stn_row = df_stations[df_stations["station"].str.lower() == station_name.lower()]
    if len(stn_row) == 0:
        print(f"  ⚠️ No coordinates found for {station_name}, skipping!")
        continue

    st_lat = float(stn_row["latitude"].iloc[0])
    st_lon = float(stn_row["longitude"].iloc[0])

    # Get this station's air quality data
    stn_aq = df_aq[df_aq["station"] == station_name].copy()
    stn_aq = stn_aq.sort_values("datetime").reset_index(drop=True)
    print(f"  Coordinates: ({st_lat:.4f}, {st_lon:.4f})")
    print(f"  AQ records: {len(stn_aq):,}")
    print(f"  Date range: {stn_aq['datetime'].min()} → {stn_aq['datetime'].max()}")

    # Compute fire features for each hourly observation
    fire_feature_list = []
    start_time = time.time()

    for idx, row in stn_aq.iterrows():
        t_end = row["datetime"]
        t_start = t_end - pd.Timedelta(hours=LOOKBACK_HOURS)
        wind_dir = row.get("wind_direction", np.nan)
        if pd.isna(wind_dir):
            wind_dir = np.nan

        # Get fires in 24-hour lookback window
        f_lats, f_lons, f_powers = get_fires_in_window(t_start, t_end)

        # Compute all spatial fire features
        features = compute_fire_features_for_window(
            f_lats, f_lons, f_powers, st_lat, st_lon, wind_dir
        )
        fire_feature_list.append(features)

        # Progress reporting
        if (idx + 1) % 5000 == 0:
            elapsed = time.time() - start_time
            rate = (idx + 1) / elapsed
            remaining = (len(stn_aq) - idx - 1) / rate / 60
            print(f"    Row {idx+1:,}/{len(stn_aq):,} | "
                  f"{rate:.0f} rows/sec | ~{remaining:.1f} min remaining")

    # Convert fire features to DataFrame and merge
    df_fire = pd.DataFrame(fire_feature_list)
    stn_aq = pd.concat([stn_aq.reset_index(drop=True), df_fire], axis=1)

    # Save to feature store
    output_path = f"data/feature_store/{station_name}_with_fire.parquet"
    stn_aq.to_parquet(output_path, index=False)

    elapsed_total = (time.time() - start_time) / 60
    print(f"  ✓ Completed in {elapsed_total:.1f} minutes")
    print(f"  ✓ Saved: {output_path}\n")

print("=" * 60)
print("✅ ALL STATIONS FIRE FEATURES COMPUTED!")
print("=" * 60)
```

Load each station's fire-feature file and add ALL remaining features:

PHASE 4 — Temporal Fire Accumulations:
  frp_sum_6h/12h/24h/48h: Rolling sum of FRP over different windows
  fires_sum_24h: Rolling fire count in 24h
  WHY: Smoke accumulates. 48 hours of continuous burning produces
  worse air quality than a single-hour fire at the same intensity.

PHASE 5 — PM2.5 Memory Features (STRONGEST PREDICTORS):
  Lags: pm25_lag1/3/6/12/24 — PM2.5 is highly autocorrelated
  Rolling means: pm25_mean_6h/12h/24h — captures persistent episodes
  Rolling max: pm25_max_24h — captures worst recent conditions
  FRP lags: frp_lag1/3/6/12/24 — temporal memory of fire activity
  WHY: Current PM2.5 is the best predictor of next-hour PM2.5.
  These features capture "smoke memory."

PHASE 6 — Weather Features:
  wind_u, wind_v — decomposition from speed + direction
  WHY: Wind direction is circular (350° ≈ 10°). Decomposing into
  U/V components makes it linear and ML-friendly.

PHASE 7 — Seasonal Features:
  hour, month, Australian season (Summer = Dec–Feb)
  sin/cos cyclic encoding for hour and month
  WHY: Hour 23 is close to hour 0 numerically they're far apart.
  Sine/cosine encoding preserves this cyclical relationship.

PHASE 8 — Target Variable:
  target_pm25 = PM2.5 at (t+1) — the NEXT hour's PM2.5
  WHY: We predict FUTURE air quality. Using current PM2.5 as both
  feature and target = data leakage.


```python
def get_australian_season(month):
    """Returns Australian season name from month number."""
    if month in [12, 1, 2]:
        return "Summer"
    elif month in [3, 4, 5]:
        return "Autumn"
    elif month in [6, 7, 8]:
        return "Winter"
    else:
        return "Spring"


def engineer_all_features(df):
    """
    Apply all feature engineering phases to a single station DataFrame.
    Input must already have fire spatial features computed.
    Returns the fully engineered DataFrame.
    """
    df = df.sort_values("datetime").reset_index(drop=True)

    # ---- PHASE 4: Temporal Fire Accumulations ----
    df["frp_sum_6h"] = df["frp_500km"].rolling(6, min_periods=1).sum()
    df["frp_sum_12h"] = df["frp_500km"].rolling(12, min_periods=1).sum()
    df["frp_sum_24h"] = df["frp_500km"].rolling(24, min_periods=1).sum()
    df["frp_sum_48h"] = df["frp_500km"].rolling(48, min_periods=1).sum()
    df["fires_sum_24h"] = df["fires_500km"].rolling(24, min_periods=1).sum()

    # ---- PHASE 5: PM2.5 Memory + FRP Lags ----
    for lag in [1, 3, 6, 12, 24]:
        df[f"pm25_lag{lag}"] = df["pm25"].shift(lag)
        df[f"frp_lag{lag}"] = df["frp_500km"].shift(lag)

    df["pm25_mean_6h"] = df["pm25"].rolling(6, min_periods=1).mean()
    df["pm25_mean_12h"] = df["pm25"].rolling(12, min_periods=1).mean()
    df["pm25_mean_24h"] = df["pm25"].rolling(24, min_periods=1).mean()
    df["pm25_max_24h"] = df["pm25"].rolling(24, min_periods=1).max()

    # ---- PHASE 6: Weather Features ----
    wind_dir_rad = np.radians(df["wind_direction"].fillna(0))
    wind_spd = df["wind_speed"].fillna(0)
    df["wind_u"] = wind_spd * np.cos(wind_dir_rad)
    df["wind_v"] = wind_spd * np.sin(wind_dir_rad)

    # ---- PHASE 7: Seasonal Features ----
    df["hour"] = df["datetime"].dt.hour
    df["month"] = df["datetime"].dt.month
    df["season"] = df["month"].apply(get_australian_season)
    df["sin_hour"] = np.sin(2 * np.pi * df["hour"] / 24.0)
    df["cos_hour"] = np.cos(2 * np.pi * df["hour"] / 24.0)
    df["sin_month"] = np.sin(2 * np.pi * df["month"] / 12.0)
    df["cos_month"] = np.cos(2 * np.pi * df["month"] / 12.0)

    # ---- PHASE 8: Target Variable ----
    # target_pm25 = PM2.5 at (t+1) — prevents data leakage
    df["target_pm25"] = df["pm25"].shift(-1)

    return df


# Process each station
for station_name in station_list:
    input_path = f"data/feature_store/{station_name}_with_fire.parquet"

    if not os.path.exists(input_path):
        print(f"  Skipping {station_name} — fire features not found")
        continue

    print(f"Engineering features for {station_name}...")

    df_stn = pd.read_parquet(input_path)
    df_stn = engineer_all_features(df_stn)

    # Drop rows where critical features are NaN (from lag/rolling calculations)
    df_stn_clean = df_stn.dropna(subset=["pm25_lag24", "target_pm25"]).reset_index(drop=True)

    output_path = f"data/feature_store/{station_name}_features.parquet"
    df_stn_clean.to_parquet(output_path, index=False)

    print(f"  ✓ {station_name}: {len(df_stn):,} → {len(df_stn_clean):,} rows | "
          f"{len(df_stn_clean.columns)} features")

print("\n✅ All stations fully feature-engineered!")
```

MULTI-HORIZON PM2.5 FORECASTING
=================================
Predict PM2.5 at 1h, 6h, 12h, 24h ahead using gradient boosted trees.

CORRECT FORECASTING APPROACH:
  At forecast time t, we predict PM2.5 at time (t + h) for horizon h.
  
  ALL features from time t are available at forecast time:
    - pm25_lag1 at time t = PM2.5 measured at t-1  → known at time t  ✓
    - pm25_lag24 at time t = PM2.5 measured at t-24 → known at time t ✓
    - fire features at time t → computed from hotspots up to time t   ✓
    - weather at time t → observed at time t                          ✓
  
  The TARGET changes with horizon:
    target_h1  = PM2.5 at t+1
    target_h6  = PM2.5 at t+6
    target_h12 = PM2.5 at t+12
    target_h24 = PM2.5 at t+24
  
  We keep ALL features because they represent the CURRENT state at
  the moment we make the forecast. The model must learn to predict
  further into the future using the same current information.

PERSISTENCE BASELINE:
  Persistence at horizon h: predict PM2.5(t+h) = PM2.5(t)
  This is the simplest possible forecast — assume nothing changes.
  It's very strong at 1h but fails badly at 24h because conditions
  change substantially over a full day.

WHY BOTH LightGBM AND XGBoost:
  The project brief requires comparing multiple ML approaches.
  Both are gradient boosted trees but differ in tree-building:
    - LightGBM: leaf-wise growth, histogram-based (faster training)
    - XGBoost: level-wise growth, different regularisation strategy
  One may outperform the other depending on data characteristics.
  Comparing both demonstrates thorough model selection.

EXPECTED RESULTS:
  - At 1h: persistence is strong; ML models slightly better
  - At 6–24h: persistence degrades rapidly; ML models maintain
    performance by leveraging fire features and wind patterns
    that capture upcoming smoke transport
  - LightGBM advantage should be LARGEST at longer horizons


```python
HORIZONS = [1, 6, 12, 24]
s2_results = []

print("=" * 70)
print("  STUDENT 2: MULTI-HORIZON SPATIOTEMPORAL FORECASTING")
print("=" * 70)

for horizon in HORIZONS:
    print(f"\n{'─' * 60}")
    print(f"  HORIZON: {horizon}h ahead")
    print(f"{'─' * 60}")

    # ------------------------------------------------------------------
    # Create target: PM2.5 at (t + horizon), shifted PER STATION
    # to avoid leaking data across station boundaries
    # ------------------------------------------------------------------
    df_all[f"target_h{horizon}"] = df_all.groupby("station")["pm25"].shift(-horizon)

    # ALL features from time t are used — no removal needed
    features_to_use = FEATURE_COLS_WITH_STATION

    # Temporal split: train 2016-2018, test 2019-2020
    df_tr = df_all[df_all["datetime"] <= TRAIN_END].dropna(
        subset=features_to_use + [f"target_h{horizon}"]
    )
    df_te = df_all[
        (df_all["datetime"] >= TEST_START) & (df_all["datetime"] <= TEST_END)
    ].dropna(subset=features_to_use + [f"target_h{horizon}"])

    X_tr = df_tr[features_to_use]
    y_tr = df_tr[f"target_h{horizon}"]
    X_te = df_te[features_to_use]
    y_te = df_te[f"target_h{horizon}"]

    print(f"  Train: {len(X_tr):,} samples | Test: {len(X_te):,} samples | "
          f"Features: {len(features_to_use)}")

    # ------------------------------------------------------------------
    # MODEL 1: LightGBM
    # ------------------------------------------------------------------
    lgb = LGBMRegressor(
        n_estimators=500, learning_rate=0.05, max_depth=8,
        subsample=0.8, colsample_bytree=0.8, min_child_samples=20,
        random_state=42, verbose=-1, n_jobs=-1
    )
    lgb.fit(X_tr, y_tr)
    yp_lgb = lgb.predict(X_te)

    mae_lgb = mean_absolute_error(y_te, yp_lgb)
    rmse_lgb = np.sqrt(mean_squared_error(y_te, yp_lgb))
    r2_lgb = r2_score(y_te, yp_lgb)

    s2_results.append({
        "Horizon": f"{horizon}h", "Model": "LightGBM",
        "MAE": mae_lgb, "RMSE": rmse_lgb, "R²": r2_lgb
    })
    print(f"  LightGBM:    MAE={mae_lgb:.3f}, RMSE={rmse_lgb:.3f}, R²={r2_lgb:.3f}")

    # ------------------------------------------------------------------
    # MODEL 2: XGBoost
    # ------------------------------------------------------------------
    xgb = XGBRegressor(
        n_estimators=500, learning_rate=0.05, max_depth=8,
        subsample=0.8, colsample_bytree=0.8,
        random_state=42, n_jobs=-1
    )
    xgb.fit(X_tr, y_tr, eval_set=[(X_te, y_te)], verbose=False)
    yp_xgb = xgb.predict(X_te)

    mae_xgb = mean_absolute_error(y_te, yp_xgb)
    rmse_xgb = np.sqrt(mean_squared_error(y_te, yp_xgb))
    r2_xgb = r2_score(y_te, yp_xgb)

    s2_results.append({
        "Horizon": f"{horizon}h", "Model": "XGBoost",
        "MAE": mae_xgb, "RMSE": rmse_xgb, "R²": r2_xgb
    })
    print(f"  XGBoost:     MAE={mae_xgb:.3f}, RMSE={rmse_xgb:.3f}, R²={r2_xgb:.3f}")

    # ------------------------------------------------------------------
    # BASELINE: Persistence — PM2.5(t+h) = PM2.5(t)
    # ------------------------------------------------------------------
    y_persist = df_te["pm25"].values  # current PM2.5 at forecast time t
    valid = ~np.isnan(y_persist) & ~np.isnan(y_te.values)

    if valid.sum() > 0:
        mae_p = mean_absolute_error(y_te[valid], y_persist[valid])
        rmse_p = np.sqrt(mean_squared_error(y_te[valid], y_persist[valid]))
        r2_p = r2_score(y_te[valid], y_persist[valid])

        s2_results.append({
            "Horizon": f"{horizon}h", "Model": "Persistence",
            "MAE": mae_p, "RMSE": rmse_p, "R²": r2_p
        })
        print(f"  Persistence: MAE={mae_p:.3f}, RMSE={rmse_p:.3f}, R²={r2_p:.3f}")

        # Show best ML improvement over persistence
        best_ml_mae = min(mae_lgb, mae_xgb)
        best_ml_name = "LightGBM" if mae_lgb <= mae_xgb else "XGBoost"
        if mae_p > 0:
            improvement = (mae_p - best_ml_mae) / mae_p * 100
            marker = "📈" if improvement > 0 else "📉"
            print(f"  {marker} Best ML ({best_ml_name}) "
                  f"{'improves' if improvement > 0 else 'worse'} "
                  f"by {abs(improvement):.1f}% MAE vs persistence")

# ------------------------------------------------------------------
# Clean up temporary columns
# ------------------------------------------------------------------
for h in HORIZONS:
    df_all.drop(columns=[f"target_h{h}"], errors="ignore", inplace=True)

# ------------------------------------------------------------------
# Summary table
# ------------------------------------------------------------------
df_s2_results = pd.DataFrame(s2_results).round(4)

print("\n" + "=" * 70)
print("  MULTI-HORIZON RESULTS SUMMARY")
print("=" * 70)
print(df_s2_results.to_string(index=False))

# Best model per horizon
print("\n  Best model per horizon:")
for h in HORIZONS:
    h_data = df_s2_results[df_s2_results["Horizon"] == f"{h}h"]
    best = h_data.loc[h_data["MAE"].idxmin()]
    print(f"    {h}h → {best['Model']:12s}  MAE={best['MAE']:.3f}  R²={best['R²']:.3f}")
```

DEEP LEARNING SEQUENCE AND SPATIOTEMPORAL MODELS
==================================================

Model progression (as required by the project brief):
  Persistence → LightGBM → XGBoost → LSTM → TCN → ConvLSTM

Three deep learning architectures:

1. LSTM (Long Short-Term Memory)
   - Input: (batch, 24 timesteps, N features) — single station
   - Learns temporal dependencies through gated memory cells
   - Captures how PM2.5 evolves over the past 24 hours

2. TCN (Temporal Convolutional Network)
   - Input: (batch, 24 timesteps, N features) — single station
   - Dilated causal convolutions: [1, 2, 4, 8]
   - Efficiently captures long-range temporal patterns
   - Often faster to train than LSTM

3. ConvLSTM (Convolutional LSTM)
   - Input: (batch, 24 timesteps, 1, n_stations, N features) — ALL stations
   - Convolution operates ACROSS STATIONS at each timestep
   - LSTM operates ACROSS TIMESTEPS
   - Learns BOTH spatial (cross-station) AND temporal patterns
   
   WHY ConvLSTM:
     ConvLSTM was designed for gridded spatial data (satellite images).
     With 6 point stations (not a regular grid), we adapt it by treating
     stations as a 1D spatial dimension. The convolution kernel slides
     across stations, learning which station combinations predict smoke
     transport. For example, if Earlwood PM2.5 spikes predict Liverpool
     PM2.5 spikes 2 hours later (due to prevailing wind), ConvLSTM
     captures this spatial propagation pattern.

   ADAPTATION:
     Original ConvLSTM expects: (time, height, width, channels) — 2D grid
     Our adaptation uses:       (time, 1, n_stations, channels) — 1D spatial
     The (1,3) kernel convolves across 3 stations at each timestep.

NOTE: LSTM & TCN demonstrated on one station (Earlwood).
      ConvLSTM uses ALL stations (true spatiotemporal).
      Full multi-station deep learning would need GPU for production.

```python
# ==================================================================
# PART 2: ConvLSTM (Multi-Station Spatiotemporal)
# ==================================================================
print(f"\n{'=' * 70}")
print(f"  PART 2: ConvLSTM — Multi-Station Spatiotemporal")
print(f"{'=' * 70}")

"""
ConvLSTM Input Construction:
  1. For each common datetime across all stations, create a spatial
     "frame" with shape (1, n_stations, n_features)
  2. Stack 24 consecutive frames into one sequence
  3. Final input: (batch, 24, 1, n_stations, n_features)

  The Conv kernel (1,3) slides across 3 stations at each timestep.
  The LSTM component recurs across the 24 timesteps.
"""

# Get all stations with data
stations_with_data = sorted(df_all["station"].unique())
n_stations = len(stations_with_data)
print(f"  Stations: {n_stations} — {stations_with_data}")

# Use a compact feature set for ConvLSTM (reduce dimensionality)
CONVLSTM_FEATURES = [
    "pm25", "temp", "humidity", "wind_speed", "wind_u", "wind_v",
    "nearest_fire_km", "fires_200km", "frp_200km", "frp_500km",
    "fire_exposure", "upwind_exposure",
    "frp_sum_24h", "pm25_lag1", "pm25_lag6", "pm25_lag24",
    "pm25_mean_6h", "pm25_mean_24h",
]
n_features_conv = len(CONVLSTM_FEATURES)
print(f"  Features per station: {n_features_conv}")
print(f"  Tensor shape per sample: ({WINDOW_SIZE}, 1, {n_stations}, {n_features_conv})")
```

```python
# ------------------------------------------------------------------
# Build aligned multi-station time series
# ------------------------------------------------------------------
# For each station, extract features indexed by datetime
station_dfs = {}
for stn in stations_with_data:
    sdf = (df_all[df_all["station"] == stn]
           [["datetime"] + CONVLSTM_FEATURES + [TARGET_COL]]
           .copy()
           .sort_values("datetime")
           .set_index("datetime"))
    station_dfs[stn] = sdf

# Find common datetimes where ALL stations have data
common_index = station_dfs[stations_with_data[0]].index
for stn in stations_with_data[1:]:
    common_index = common_index.intersection(station_dfs[stn].index)

common_index = common_index.sort_values()
print(f"  Common datetime hours across all stations: {len(common_index):,}")

if len(common_index) > WINDOW_SIZE + 500:

    # Build spatial tensor: (n_timesteps, n_stations, n_features)
    spatial_features = np.zeros((len(common_index), n_stations, n_features_conv))
    for i, stn in enumerate(stations_with_data):
        stn_data = station_dfs[stn].loc[common_index, CONVLSTM_FEATURES]
        spatial_features[:, i, :] = stn_data.values

    # Target: first station's PM2.5 (could also use average)
    # Using first station for consistency with LSTM/TCN comparison
    target_station = stations_with_data[0]
    spatial_targets = (station_dfs[target_station]
                       .loc[common_index, TARGET_COL]
                       .values.reshape(-1, 1))

```

```python

    # ------------------------------------------------------------------
    # Handle NaN: drop rows where any value is NaN
    # ------------------------------------------------------------------
    valid_mask = (~np.isnan(spatial_features).any(axis=(1, 2)) &
                  ~np.isnan(spatial_targets).ravel())

    spatial_features_clean = spatial_features[valid_mask]
    spatial_targets_clean = spatial_targets[valid_mask]
    common_index_clean = common_index[valid_mask]

    print(f"  After NaN removal: {len(spatial_features_clean):,} timesteps")

    # ------------------------------------------------------------------
    # Scale features
    # ------------------------------------------------------------------
    scaler_conv_x = StandardScaler()
    scaler_conv_y = StandardScaler()

    # Reshape for scaling: (time, stations*features)
    n_time = len(spatial_features_clean)
    flat_features = spatial_features_clean.reshape(n_time, -1)
    flat_scaled = scaler_conv_x.fit_transform(flat_features)
    spatial_scaled = flat_scaled.reshape(n_time, n_stations, n_features_conv)

    targets_scaled = scaler_conv_y.fit_transform(spatial_targets_clean)

    # ------------------------------------------------------------------
    # Create ConvLSTM sequences
    # ------------------------------------------------------------------
    def create_conv_sequences(features, targets, window_size=24):
        """
        Create ConvLSTM sequences.

        Input:
          features: (n_time, n_stations, n_features)
          targets:  (n_time, 1)

        Output:
          X: (n_samples, window_size, 1, n_stations, n_features)
          y: (n_samples, 1)

        The extra dimension (1) is the "height" for ConvLSTM2D.
        ConvLSTM2D expects 5D input: (batch, time, rows, cols, channels)
        We use rows=1, cols=n_stations, channels=n_features
        """
        X, y = [], []
        for i in range(len(features) - window_size):
            # Shape: (window, stations, features) → (window, 1, stations, features)
            seq = features[i:i + window_size]
            seq = seq[:, np.newaxis, :, :]  # add height=1 dimension
            X.append(seq)
            y.append(targets[i + window_size])
        return np.array(X), np.array(y)

    X_conv, y_conv = create_conv_sequences(spatial_scaled, targets_scaled, WINDOW_SIZE)

    # Temporal split for ConvLSTM
    dt_conv = common_index_clean[WINDOW_SIZE:]
    tr_mask_conv = dt_conv <= pd.Timestamp(TRAIN_END)
    te_mask_conv = ((dt_conv >= pd.Timestamp(TEST_START)) &
                    (dt_conv <= pd.Timestamp(TEST_END)))

    X_train_conv = X_conv[tr_mask_conv]
    y_train_conv = y_conv[tr_mask_conv]
    X_test_conv = X_conv[te_mask_conv]
    y_test_conv = y_conv[te_mask_conv]
    y_test_conv_actual = scaler_conv_y.inverse_transform(y_test_conv)

    print(f"  ConvLSTM input shape: {X_conv.shape}")
    print(f"  Train: {len(X_train_conv):,} | Test: {len(X_test_conv):,}")

    if len(X_train_conv) > 100 and len(X_test_conv) > 100:

        # ----------------------------------------------------------
        # Build ConvLSTM Model
        # ----------------------------------------------------------
        print("\n  Training ConvLSTM...")
        print(f"  Architecture: ConvLSTM2D(32) → ConvLSTM2D(16) → Dense(32) → Dense(1)")
        print(f"  Kernel: (1,3) — slides across 3 stations at each timestep")

        conv_input = Input(shape=(WINDOW_SIZE, 1, n_stations, n_features_conv))

        # First ConvLSTM layer: learn spatial-temporal patterns
        x = ConvLSTM2D(
            filters=32,
            kernel_size=(1, 3),         # convolve across 3 stations
            padding="same",             # preserve spatial dimension
            return_sequences=True,      # pass full sequence to next layer
            activation="relu",
        )(conv_input)
        x = BatchNormalization()(x)
        x = Dropout(0.2)(x)

        # Second ConvLSTM layer: deeper spatiotemporal features
        x = ConvLSTM2D(
            filters=16,
            kernel_size=(1, 3),
            padding="same",
            return_sequences=False,     # only final timestep output
            activation="relu",
        )(x)
        x = BatchNormalization()(x)
        x = Dropout(0.2)(x)

        # Flatten spatial output and predict
        x = Flatten()(x)
        x = Dense(32, activation="relu")(x)
        x = Dropout(0.2)(x)
        output = Dense(1)(x)

        convlstm_model = Model(inputs=conv_input, outputs=output)
        convlstm_model.compile(optimizer="adam", loss="mse")

        print(f"  Total parameters: {convlstm_model.count_params():,}")

        # Train
        history_conv = convlstm_model.fit(
            X_train_conv, y_train_conv,
            epochs=30, batch_size=32,
            validation_split=0.1, verbose=0,
            callbacks=[early_stop]
        )

        # Predict and evaluate
        preds_conv_scaled = convlstm_model.predict(X_test_conv, verbose=0)
        preds_conv = scaler_conv_y.inverse_transform(preds_conv_scaled)

        mae_conv = mean_absolute_error(y_test_conv_actual, preds_conv)
        rmse_conv = np.sqrt(mean_squared_error(y_test_conv_actual, preds_conv))
        r2_conv = r2_score(y_test_conv_actual, preds_conv)

        dl_results.append({
            "Model": "ConvLSTM", "Type": "Multi-station (spatiotemporal)",
            "MAE": mae_conv, "RMSE": rmse_conv, "R²": r2_conv
        })
        print(f"  ✓ ConvLSTM: MAE={mae_conv:.3f}, RMSE={rmse_conv:.3f}, R²={r2_conv:.3f}")

        # Training history plot
        fig, ax = plt.subplots(figsize=(8, 4))
        ax.plot(history_conv.history["loss"], label="Train Loss")
        ax.plot(history_conv.history["val_loss"], label="Val Loss")
        ax.set_title("ConvLSTM Training History")
        ax.set_xlabel("Epoch"); ax.set_ylabel("MSE Loss")
        ax.legend(); ax.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig("S2_convlstm_training.png", dpi=150, bbox_inches="tight")
        plt.show()

    else:
        print("  ⚠️ Not enough aligned data for ConvLSTM after splitting.")

else:
    print("  ⚠️ Not enough common datetime hours across stations for ConvLSTM.")
    print(f"     Need >{WINDOW_SIZE + 500}, have {len(common_index)}")
```

```python
print("=" * 70)
print("  STUDENT 2: SMOKE PROPAGATION ANALYSIS (Black Summer)")
print("=" * 70)

BS_START = "2019-10-01"
BS_END = "2020-03-31"

bs_data = df_all[
    (df_all["datetime"] >= BS_START) & (df_all["datetime"] <= BS_END)
].copy()

# Pivot: each station's PM2.5 as a separate time series column
pm25_pivot = bs_data.pivot_table(
    index="datetime", columns="station", values="pm25", aggfunc="mean"
)

print(f"  Black Summer period: {BS_START} → {BS_END}")
print(f"  Stations with data: {list(pm25_pivot.columns)}")
print(f"  Time points: {len(pm25_pivot):,}")
```

```python
# ------------------------------------------------------------------
# Cross-correlation with lag analysis
# ------------------------------------------------------------------
max_lag_hours = 24
propagation_results = []

for stn_a, stn_b in combinations(pm25_pivot.columns, 2):
    sa = pm25_pivot[stn_a].dropna()
    sb = pm25_pivot[stn_b].dropna()
    common = sa.index.intersection(sb.index)

    if len(common) < 100:
        continue

    a = sa.loc[common].values
    b = sb.loc[common].values

    # Find lag with maximum correlation
    best_corr, best_lag = 0, 0
    for lag in range(-max_lag_hours, max_lag_hours + 1):
        if lag >= 0:
            if len(a) > lag + 10:
                c = np.corrcoef(a[lag:], b[:len(b) - lag])[0, 1]
            else:
                c = 0
        else:
            if len(b) > -lag + 10:
                c = np.corrcoef(a[:len(a) + lag], b[-lag:])[0, 1]
            else:
                c = 0

        if not np.isnan(c) and abs(c) > abs(best_corr):
            best_corr = c
            best_lag = lag

    propagation_results.append({
        "Station A": stn_a,
        "Station B": stn_b,
        "Zero-lag Corr": round(np.corrcoef(a, b)[0, 1], 4),
        "Max Corr": round(best_corr, 4),
        "Best Lag (h)": best_lag,
    })

df_propagation = pd.DataFrame(propagation_results)
print("\nStation-Pair Smoke Propagation (Black Summer):")
print(df_propagation.to_string(index=False))

print("\nInterpretation:")
print("  • Lag > 0: Station A leads (smoke arrives at A first)")
print("  • Lag < 0: Station B leads (smoke arrives at B first)")
print("  • Lag = 0: Simultaneous (regional smoke event)")
print("  • Higher correlation = more coupled smoke exposure")
print("✅ Saved: S2_smoke_propagation.png")
```
